```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from plotly.subplots import make_subplots
import plotly.express as px
import plotly.graph_objects as go

path = os.getcwd()
DATA_PATH = os.path.dirname(os.path.dirname(path)) + '\\data'
```

```python
RNAp_df = pd.read_csv(DATA_PATH + '\\RNAp_with_Raw_PCN.csv')
RNAi_df = pd.read_csv(DATA_PATH+ '\\RNAi_with_Raw_PCN.csv')

RNAp_predicted_PCN = RNAp_df['Copy Number']
RNAp_raw_PCN = RNAp_df['Raw Copy Number']

RNAi_predicted_PCN = RNAi_df['Copy Number']
RNAi_raw_PCN = RNAi_df['Raw Copy Number']
```

```python
fig1 = px.histogram(RNAp_raw_PCN, title='RNAp Raw Copy Number histogram',x="Raw Copy Number")
fig1.show()
```

```python
fig2 = px.histogram(np.log(RNAp_raw_PCN), title='RNAp log Raw Copy Number histogram',x="Raw Copy Number")
fig2.show()
```

```python
fig3 = px.histogram(RNAi_raw_PCN, title='RNAi Raw Copy Number histogram',x="Raw Copy Number")
fig3.show()
```

```python
fig4 = px.histogram(np.log(RNAi_raw_PCN), title='RNAi log Raw Copy Number histogram',x="Raw Copy Number")
fig4.show()
```

```python
fig = go.Figure(data=[go.Histogram(x=RNAp_raw_PCN, name='RNAp')])
fig.add_trace(go.Histogram(x=RNAi_raw_PCN, name='RNAi'))
fig.update_layout(barmode='overlay', title='Raw Copy Number histogram')
```

```python
fig = go.Figure(data=[go.Histogram(x=np.log(RNAp_raw_PCN), name='RNAp')])
fig.add_trace(go.Histogram(x=np.log(RNAi_raw_PCN), name='RNAi'))
fig.update_layout(barmode='overlay', title='Log Raw Copy Number histogram')
```

```python
fig = make_subplots(rows=2, cols=2)
fig.add_trace(go.Histogram(x=RNAp_raw_PCN, name='RNAp'), row=1, col=1)
fig.add_trace(go.Histogram(x=RNAi_raw_PCN, name='RNAi'), row=1, col=2)
fig.add_trace(go.Histogram(x=np.log(RNAp_raw_PCN), name='RNAp log'), row=2, col=1)
fig.add_trace(go.Histogram(x=np.log(RNAi_raw_PCN), name='RNAi log'), row=2, col=2)
fig.update_layout(title='Raw Copy Number histogram', title_x=0.5)

```

```python
fig.write_html(path + '\\raw_data_histograms.html')
```
